```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv('Food_Delivery_Time_Prediction.csv')

# Central Tendencies
print("Mean Delivery Time:", df['Time_taken_min'].mean())
print("Median Delivery Time:", df['Time_taken_min'].median())

# Correlation Matrix
plt.figure(figsize=(10, 6))
numeric_cols = ['Order_Items', 'Road_Distance_km', 'Preparation_Time_Min', 'Time_taken_min']
corr_matrix = df[numeric_cols].corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap of Logistics Variables')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x='Traffic_Level', y='Time_taken_min', data=df, palette='Set2',
            order=['Low', 'Moderate', 'High', 'Severe'])
plt.title('Delivery Time Distribution by Traffic Level')
plt.xlabel('Traffic Level')
plt.ylabel('Time Taken (Minutes)')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Road_Distance_km', y='Time_taken_min', hue='Weather',
                data=df.sample(1000), palette='viridis', alpha=0.7)
plt.title('Road Distance vs. Time Taken (Colored by Weather)')
plt.xlabel('Road Distance (km)')
plt.ylabel('Time Taken (min)')
plt.legend(title='Weather')
plt.show()
```
